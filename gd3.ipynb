# B4: Implement Gradient Descent to find a local minimum of y = (x + 3)² starting at x = 2.

```python
def f(x):
    return (x+3)**2

def df(x):
    return 2*(x+3)
```

```python
gd = []
func = []

def gradient_descent(start, learning_rate, n_iterations):
    x = start
    gd.append(x)
    func.append(f(x))
    for i in range(n_iterations):
        x = x - learning_rate*df(x)
        gd.append(x)
        func.append(f(x))
        print(f"Itertaion {i}: x = {x}, f(x) = {f(x)}")
    return x
```

```python
start = 2
learning_rate = 0.1
n_iterations = 100

gradient_descent(start=start, learning_rate=learning_rate, n_iterations=n_iterations)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.scatterplot(data=gd)
plt.title('`x` vs Iterations')
plt.xlabel('Iterations')
plt.ylabel('`x`')
plt.show()

sns.scatterplot(data=func)
plt.title('`f(x)` vs Iterations')
plt.xlabel('Iterations')
plt.ylabel('`f(x)`')
plt.show()
```
